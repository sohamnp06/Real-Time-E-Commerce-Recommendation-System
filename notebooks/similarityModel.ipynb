```python
import pandas as pd
import pickle
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv("../data/amazonSale.csv", encoding="latin-1")
```

```python
customer_product_matrix = df.pivot_table(
    index="Customer ID",
    columns="Product ID",
    values="Quantity",
    aggfunc="sum",
    fill_value=0
)
```

```python
product_customer_matrix = customer_product_matrix.T
```

```python
product_similarity = cosine_similarity(product_customer_matrix)
```

```python
product_similarity_df = pd.DataFrame(
    product_similarity,
    index=product_customer_matrix.index,
    columns=product_customer_matrix.index
)
```

```python
def recommend_similar_products(product_id, n=5):

    similar_products = product_similarity_df[product_id] \
                        .sort_values(ascending=False)

    return similar_products.iloc[1:n+1]
```

```python
recommend_similar_products("FUR-BO-10001798")
```

```python
import pickle

with open("../files/product_similarity_matrix.pkl", "wb") as f:
    pickle.dump(product_similarity_df, f)
```
